```python
# Convert the HTML text into a BeautifulSoup object
soup = bs4.BeautifulSoup(request1.text, "html.parser")

# Print parsed HTML
print(soup)
```

```python

product_name = soup.find("span", {"id": "productTitle"})


if product_name:
    product_name = product_name.get_text(strip=True)

print(product_name)
```

```python
# Amazon price is usually inside span with class "a-offscreen"
price = soup.select_one("span.a-price span.a-offscreen")

# Clean and print price
if price:
    price = price.get_text(strip=True)

print(price)
```

```python
# Amazon average customer rating is stored inside averageCustomerReviews
rating = soup.select_one("#averageCustomerReviews .a-icon-alt")

# Clean and print rating
if rating:
    rating = rating.get_text(strip=True)

print(rating)
```

```python
# Amazon review count is stored inside id "acrCustomerReviewText"
review_count = soup.select_one("#acrCustomerReviewText")

# Clean and print review count
if review_count:
    review_count = review_count.get_text(strip=True)

print(review_count)
```

```python
# Each Amazon review is stored inside a list item with data-hook="review"
review_blocks = soup.select('li[data-hook="review"]')

# Print how many review blocks were found on the current page
print("Total reviews found on this page:", len(review_blocks))
```

```python
# Loop through each review block and extract customer name
for review in review_blocks:
    customer_name = review.select_one(".a-profile-name")

    if customer_name:
        print(customer_name.get_text(strip=True))
```

```python
# Loop through each review block and extract review rating
for review in review_blocks:
    review_rating = (
        review.select_one('i[data-hook="review-star-rating"] span')
        or review.select_one('i[data-hook="cmps-review-star-rating"] span')
        or review.select_one("i.review-rating span.a-icon-alt")
    )

    if review_rating:
        print(review_rating.get_text(strip=True))
```

```python
# Loop through each review block and extract review title
for review in review_blocks:
    review_title = review.select_one('[data-hook="review-title"]')

    if review_title:
        print(review_title.get_text(" ", strip=True))
```
